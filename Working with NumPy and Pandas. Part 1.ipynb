# Прочитай набір даних із CSV-файлу та виконай очищення даних — видали усі рядки, які містять символ "?" у будь-яких стовпцях. Також перевір коректність даних (зарплата та зарплата у $K).

```python
adult_data["check_salary"] = (
    ((adult_data["salary"] == "<=50K") & (adult_data["salary K$"] <= 50)) |
    ((adult_data["salary"] == ">50K") & (adult_data["salary K$"] > 50))
)

print(adult_data["check_salary"].value_counts())
```

# Перевір, чи є люди без вищої освіти (освіта: Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters, Doctorate), але із зарплатою > 50K.

```python
print(adult_data["education"].unique())
```

```python
adult_data_filtered = adult_data[
    ~adult_data["education"].isin([
        "Bachelors", "Prof-school", "Assoc-acdm",
        "Assoc-voc", "Masters", "Doctorate"
    ])
    & adult_data["salary K$"] > 50
]

print(adult_data_filtered.head())
```

# Порівняй зарплати одружених і неодружених чоловіків. Хто заробляє більше? (> 50K або < = 50K) Одружені чоловіки — це ті, у кого marital-status починається з Married. Інші — не одружені.

```python
pivot = pd.pivot_table(
    adult_data[adult_data["sex"] == "Male"],
    index=adult_data["marital-status"].str.startswith("Married"),
    columns="salary",
    values="age",
    aggfunc="count"
)

print(pivot)
```
